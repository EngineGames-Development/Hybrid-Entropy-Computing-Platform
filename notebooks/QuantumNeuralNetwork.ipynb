```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import time

df = pd.read_csv('sample_data/quantum_ai_data.csv')
df_clean = df[(df['Quantum_Decision'] >= 0) & (df['Quantum_Decision'] <= 1)]

def analyze_complexity(data):
    sizes = [10, 100, 500, 1000, 2000, len(data)]
    times = []
    for s in sizes:
        start_time = time.time()
        _ = data[:s].mean()
        _ = data[:s].std()
        times.append(time.time() - start_time)
    return sizes, times

sizes, times = analyze_complexity(df_clean['Quantum_Decision'])

plt.style.use('dark_background')
fig = plt.figure(figsize=(20, 12))
gs = fig.add_gridspec(2, 2, hspace=0.3, wspace=0.2)

fig.suptitle('QUANTUM-AI HYBRID SYSTEM | ANALYTICS TERMINAL V2.0', fontsize=22, color='cyan')

ax1 = fig.add_subplot(gs[0, 0])
ax1.plot(df_clean['Quantum_Decision'].values, color='#00ff00', linewidth=1, alpha=0.9)
ax1.set_title('Real-Time Decision Output', color='cyan')
ax1.set_ylabel('Probability State')
ax1.grid(True, alpha=0.1)

ax2 = fig.add_subplot(gs[0, 1])
ax2.hist(df_clean['Quantum_Decision'], bins=25, color='magenta', alpha=0.7, rwidth=0.85)
ax2.set_title('Entropy Distribution (Fairness)', color='magenta')
ax2.grid(True, axis='y', alpha=0.1)

ax3 = fig.add_subplot(gs[1, :])
ax3.plot(sizes, times, 'o-', color='yellow', label='Measured Execution Time')

ax3.plot(sizes, [t * (s/sizes[0]) for s, t in zip(sizes, [times[0]]*len(sizes))], '--', color='grey', alpha=0.5, label='Theoretical O(n)')
ax3.set_title('Computational Complexity Analysis (Big O Notation)', fontsize=14, color='yellow')
ax3.set_xlabel('Input Size (n Samples)')
ax3.set_ylabel('Execution Time (seconds)')
ax3.legend()
ax3.grid(True, alpha=0.1)

mean_val = df_clean['Quantum_Decision'].mean()
std_dev = df_clean['Quantum_Decision'].std()

print(f"--- QUANTUM METRICS ---")
print(f"Mathematical Average: {mean_val:.4f} (Ideal: 0.5000)")
print(f"Standarddeviation:    {std_dev:.4f}")
print(f"Complexity Model:     O(n) - Linear Scaling detected")

plt.savefig('sample_data/Quantum_Ultimate_Research.pdf')
plt.show()
```
